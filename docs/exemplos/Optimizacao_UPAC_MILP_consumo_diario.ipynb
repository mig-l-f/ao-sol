# Optimização par PV+Bateria para minimizar custo de energia

## Objectivo 

Encontrar o melhor par PV+Bateria com menor custo de energia. O menor custo de energia ocorre quando a última unidade instalada de PV ou bateria deixa de gerar poupança suficiente para pagar o seu custo anual. Custo marginal de nova unidade é maior que a poupança gerada.

## Metodologia

Optimização do despacho utilizando programação linear. Utilização de indicadores relacionados com o consumo diário
